## 2\. Visualizaciones y análisis

### Gráfico 1: Histograma de distribución del precio en USD

Representación gráfica de la frecuencia de publicación de los inmuebles según su precio en dólares, acotando la muestra a valores inferiores a USD 500\.000 para analizar la concentración central de los datos\.

```
df_precios = df_properati[df_properati['price_aprox_usd'] < 500000]

plt.figure(figsize=(9, 4))
sns.histplot(
    df_precios['price_aprox_usd'], 
    bins=40, 
    kde=True, 
    color='#005b96', 
    edgecolor='black'
)
plt.title('Distribución de Precios de Venta (USD)')
plt.xlabel('Precio (USD)')
plt.ylabel('Frecuencia')
plt.ticklabel_format(style='plain', axis='x')
plt.show()
```

📌 La distribución del precio muestra que la mayor parte de la oferta inmobiliaria se agrupa en el rango de los USD 50\.000 a USD 180\.000, disminuyendo la cantidad de propiedades en los valores superiores\.

### Gráfico 2: Dispersión entre superficie cubierta y precio por tipo de propiedad

Visualización de la relación entre la superficie cubierta en m2 y el precio de venta en USD para inmuebles de tipo Departamento, Casa y PH\.

```
df_filtrado = df_properati[
    (df_properati['property_type'].isin(['Departamento', 'Casa', 'PH'])) &
    (df_properati['surface_covered_in_m2'] < 300) &
    (df_properati['price_aprox_usd'] < 600000)
]

plt.figure(figsize=(9, 5))
sns.scatterplot(
    data=df_filtrado, 
    x='surface_covered_in_m2', 
    y='price_aprox_usd', 
    hue='property_type', 
    palette='bright', 
    alpha=0.7
)
plt.title('Relación entre Superficie Cubierta (m²) y Precio (USD)')
plt.xlabel('Superficie Cubierta (m²)')
plt.ylabel('Precio (USD)')
plt.ticklabel_format(style='plain', axis='y')
plt.show()
```

📌 Se observa una asociación positiva general donde mayor metraje cubierto se acompaña de un mayor precio\. Los departamentos se concentran en metrajes más reducidos en comparación con las casas\.

## 3\. Observaciones generales

Breve resumen del análisis: ¿Qué patrones se observan? ¿Hay calidad en los datos? ¿Qué variables resultan más interesantes?

- Patrones observados: La mayoría de los inmuebles están agrupados en los valores de entre 50\.000 y 180\.000 dólares\. Se ve una relación lógica entre el tamaño y el valor: a más metros cuadrados cubiertos, el precio sube\. Además, los departamentos suelen tener menos superficie que las casas, pero su precio por metro cuadrado termina siendo más alto\.

- Calidad de los datos: En general la base está bastante bien para arrancar\. Lo positivo es que la columna de precios no tiene datos faltantes\. Al ser información real bajada de un sitio web, es normal encontrar algunos vacíos en columnas secundarias \(como expensas o superficie total\) y varios precios o metrajes muy altos que se salen del promedio\.

- Variables más interesantes:

price\_aprox\_usd: Es el dato clave porque marca el valor en dólares de cada publicación\.

surface\_covered\_in\_m2: Es la característica física que más influye a la hora de entender los cambios en el precio\.

Analizaremos 5 de los equipos mas usados para los diferentes ejercicos y a su vez lo distribuiremos segun su nivel de dificultad\.

```
plt.figure()

top_equipment = df['Equipment'].value_counts().head(5).index
df_filtered = df[df['Equipment'].isin(top_equipment)]

sns.countplot(data=df_filtered, x='Equipment', hue='Level', palette='viridis')
plt.title('Nivel de Dificultad de Ejercicios en los 5 Equipos más Comunes')
plt.xlabel('Tipo de Equipo')
plt.ylabel('Cantidad de Ejercicios')
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

📌 Interpretación del gráfico: Se puede observar que el equipo que mas se utiliza es Body Only \(Solo el cuerpo\) y que a su vez brinda un nivel de ejercicios de nivel intermedio, en comparación con otros equipos\.

Se analiza el rango de precios de venta al público por cada categoría de producto para identificar los tickets de mayor valor\.

```
plt.figure(figsize=(10, 4))
sns.boxplot(data=df_nicolas, x='Rubro', y='Precio_Publico_ARS', palette='Blues_d')
plt.title('Distribución de Precios Públicos (ARS) por Rubro', fontsize=12, fontweight='bold')
plt.xticks(rotation=15)
plt.ylabel('Precio ($ ARS)')
plt.tight_layout()
plt.show()
```

Se evalúa el volumen medio de ventas mensuales asociadas a la disponibilidad del producto en inventario\.

```
plt.figure(figsize=(8, 4))
sns.barplot(data=df_nicolas, x='Estado_Stock', y='Ventas_Mensuales', estimator=np.mean, errorbar=None, palette='Blues_r')
plt.title('Ventas Mensuales Promedio según Estado de Stock', fontsize=12, fontweight='bold')
plt.xlabel('Estado de Stock')
plt.ylabel('Ventas Mensuales Promedio (Unidades)')
plt.tight_layout()
plt.show()
```

📌 Los productos clasificados como "Poco stock" o "Disponible" registran el mayor promedio de unidades vendidas al mes, indicando que los artículos de alta rotación corren riesgo de quiebre de inventario si no se reabastecen a tiempo\.

Cruzamos tres o más variables para observar patrones más complejos\.

Evaluamos la interacción simultánea entre el rubro, el color del producto y su nivel promedio de ventas mensuales\.

```
top_colores = df["Color"].value_counts().head(8).index
top_rubros = df["Rubro"].value_counts().head(5).index

datos_multi = df[df["Color"].isin(top_colores) & df["Rubro"].isin(top_rubros)].copy()

tabla_color_rubro = datos_multi.pivot_table(
    index="Rubro", columns="Color", values="Ventas_Mensuales", aggfunc="mean"
)

plt.figure(figsize=(11, 5))
sns.heatmap(tabla_color_rubro, annot=True, fmt=".1f", cmap="YlGnBu")
plt.title("Ventas mensuales promedio según rubro y color")
plt.xlabel("Color")
plt.ylabel("Rubro")
plt.tight_layout()
plt.show()
```

## 4\. Planteo de Hipótesis

A continuación, evaluamos cuatro hipótesis mediante el análisis estadístico y gráfico de los datos\.

###  Hipótesis 1: Los productos más caros no necesariamente son los más vendidos

Planteamos que los productos con precios públicos más altos \(cuartil superior\) no necesariamente registran un volumen de ventas promedio superior al de los productos de menor precio \(cuartil inferior\)\.

```
datos_h1 = df[["Precio_Publico_ARS", "Ventas_Mensuales"]].dropna().copy()
q1 = datos_h1["Precio_Publico_ARS"].quantile(0.25)
q3 = datos_h1["Precio_Publico_ARS"].quantile(0.75)

datos_h1["Grupo_precio"] = np.select(
    [datos_h1["Precio_Publico_ARS"] <= q1,
     datos_h1["Precio_Publico_ARS"] >= q3],
    ["Precio bajo", "Precio alto"],
    default="Precio intermedio"
)

orden_h1 = ["Precio bajo", "Precio intermedio", "Precio alto"]
promedios_h1 = (datos_h1.groupby("Grupo_precio", observed=False)["Ventas_Mensuales"]
                .mean().reindex(orden_h1).reset_index())

display(promedios_h1)
```

```
sns.barplot(data=promedios_h1, x="Grupo_precio", y="Ventas_Mensuales", order=orden_h1, palette="mako")
plt.title("Ventas mensuales promedio según grupo de precio")
plt.xlabel("Grupo de precio")
plt.ylabel("Ventas mensuales promedio")
plt.show()
```

Se dividió la distribución de precios en tres segmentos y se calculó la media muestral de ventas para cada categoría\.

\*\*Veredicto:\*\*

Al comparar los promedios de ventas entre estratos, el grupo de productos de precio alto no supera en ventas promedio al de precio bajo, demostrando que un precio más elevado no genera por sí solo un mayor volumen de unidades vendidas\.

### Hipótesis 2: Algunos colores presentan mayores ventas, incluso en distintos rubros

Planteamos que las ventas mensuales promedio varían según el color del producto, y que esta preferencia varía según el rubro comercial al que pertenece\.

```
top_colores = df["Color"].value_counts().head(8).index
top_rubros = df["Rubro"].value_counts().head(5).index

datos_h2 = df[df["Color"].isin(top_colores) & df["Rubro"].isin(top_rubros)].copy()
promedios_color = (datos_h2.groupby("Color")["Ventas_Mensuales"]
                   .mean().sort_values(ascending=False).reset_index())

display(promedios_color)
```

```
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.barplot(data=promedios_color, x="Color", y="Ventas_Mensuales", palette="viridis", ax=axes[0])
axes[0].set_title("Ventas mensuales promedio por color")
axes[0].tick_params(axis='x', rotation=35)

tabla_color_rubro = datos_h2.pivot_table(
    index="Rubro", columns="Color", values="Ventas_Mensuales", aggfunc="mean"
)
sns.heatmap(tabla_color_rubro, annot=True, fmt=".1f", cmap="YlGnBu", ax=axes[1])
axes[1].set_title("Ventas promedio por rubro y color")

plt.tight_layout()
plt.show()
```

Se seleccionaron los colores y rubros más frecuentes para calcular las ventas promedio generales por color y analizar su comportamiento cruzado por rubro mediante un mapa de calor\.

Se observan variaciones en los promedios de ventas según el color del artículo, confirmando que la efectividad de cada variante de color depende fuertemente del rubro específico\.

## Hipótesis 3: Los productos con más piezas tienen precios más altos, pero no necesariamente más ventas

Planteamos que los productos compuestos por una mayor cantidad de piezas presentan un precio público más alto, pero esto no se traduce necesariamente en mayores ventas mensuales\.

```
df["Cantidad_Piezas"] = df["Caracteristica_Piezas"].astype(str).str.extract(r"(\d+)").astype(float)

datos_h3 = df.dropna(subset=["Cantidad_Piezas", "Precio_Publico_ARS", "Ventas_Mensuales"]).copy()
resumen_piezas = (
    datos_h3.groupby("Cantidad_Piezas")
    .agg(Precio_promedio=("Precio_Publico_ARS", "mean"),
         Ventas_promedio=("Ventas_Mensuales", "mean"),
         Cantidad_productos=("Cantidad_Piezas", "size"))
    .reset_index()
    .sort_values("Cantidad_Piezas")
)

display(resumen_piezas)
```

```
fig, axes = plt.subplots(1, 2, figsize=(14, 4))

sns.lineplot(data=resumen_piezas, x="Cantidad_Piezas", y="Precio_promedio", marker="o", ax=axes[0], color="red")
axes[0].set_title("Precio promedio según cantidad de piezas")
axes[0].set_xlabel("Cantidad de piezas")
axes[0].set_ylabel("Precio promedio (ARS)")

sns.lineplot(data=resumen_piezas, x="Cantidad_Piezas", y="Ventas_promedio", marker="o", ax=axes[1], color="blue")
axes[1].set_title("Ventas promedio según cantidad de piezas")
axes[1].set_xlabel("Cantidad de piezas")
axes[1].set_ylabel("Ventas mensuales promedio")

plt.tight_layout()
plt.show()
```

A partir de la descripción del producto, se obtuvo el número de piezas y se analizó cómo se relaciona esta cantidad simultáneamente con el precio promedio y las ventas medias\.

Un mayor número de piezas se asocia directamente con un precio público promedio superior; sin embargo, las ventas mensuales promedio no muestran un incremento equivalente e incluso se estabilizan o descienden en sets de más piezas\.

### Hipótesis 4: Existe una relación entre el precio y las ventas mensuales

Planteamos que existe una relación de correlación lineal entre el precio público de un producto y sus ventas mensuales\.

```
datos_h4 = df[["Precio_Publico_ARS", "Ventas_Mensuales"]].dropna()

correlacion = datos_h4["Precio_Publico_ARS"].corr(datos_h4["Ventas_Mensuales"])
print(f"Coeficiente de correlación de Pearson (r): {correlacion:.4f}")
```

```
sns.regplot(
    data=datos_h4,
    x="Precio_Publico_ARS",
    y="Ventas_Mensuales",
    scatter_kws={"alpha": 0.35},
    line_kws={"color": "red"}
)
plt.title("Relación entre precio público y ventas mensuales")
plt.xlabel("Precio público (ARS)")
plt.ylabel("Ventas mensuales")
plt.tight_layout()
plt.show()
```

Se calculó el coeficiente de correlación de Pearson entre las variables continuas de precio y ventas, graficando un diagrama de dispersión con línea de ajuste lineal\.

El coeficiente de correlación de Pearson es prácticamente cero, lo que indica que no existe una relación lineal fuerte o directa entre el precio y la cantidad de unidades vendidas\.
