## Dataset simple con Pandas

```
datos = {
    "estudiante": ["Ana", "Juan", "Pedro", "Carla", "Lucía"],
    "nota": [8, 6, 5, 9, 7]
}

df = pd.DataFrame(datos)

promedio = df["nota"].mean()

nota_maxima = df["nota"].max()

nota_minima = df["nota"].min()

print(df)
print(("promedio:") , promedio)
print(("nota máxima:") , nota_maxima)
print(("nota mínima:") , nota_minima)
```
